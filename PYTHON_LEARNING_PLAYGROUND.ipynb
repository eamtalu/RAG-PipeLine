## 4. Deep dive: `dict`
Key → value mapping. Fast lookup by key.

```python
meta = {"profile": "cv", "page": 2}
print(meta["profile"])              # exact (KeyError if missing)
print(meta.get("missing"))          # None if missing
print(meta.get("missing", "n/a"))   # fallback
```

```python
# Build a dict conditionally, then spread it as keyword args
def fake_update(**kwargs):
    return kwargs

values = {"status": "failed"}
error = "boom"
if error:
    values["error"] = error
print(fake_update(**values))
```

```python
# Turn a list of objects into an id -> object lookup table
class Row:
    def __init__(self, id, name):
        self.id, self.name = id, name

rows = [Row(1, "a"), Row(2, "b")]
by_id = {r.id: r for r in rows}     # dict comprehension
print(by_id[2].name)
```

```python
# Merge / override with ** (later keys win)
base = {"text": "hi", "a": 1}
print({**base, "text": "bye"})
```

## 6. Comprehensions & generator expressions

## 9. Classes & OOP
`__init__`/`self`, inheritance, ABC + Strategy, `Enum`, `@dataclass`.

```python
class DataIngestion:
    def __init__(self, storage, repo):   # constructor
        self.storage = storage           # store on the instance
        self.repo = repo
    def describe(self):
        return f"{self.storage} + {self.repo}"

print(DataIngestion("local", "jobrepo").describe())
```

```python
# Abstract Base Class + Strategy + Factory (mirrors your parsers)
from abc import ABC, abstractmethod

class BaseParser(ABC):
    @abstractmethod
    def parse(self, data): ...

class PdfParser(BaseParser):
    def parse(self, data): return "pdf:" + data

class HtmlParser(BaseParser):
    def parse(self, data): return "html:" + data

REGISTRY = {"application/pdf": PdfParser, "text/html": HtmlParser}

def get_parser(mime):
    return REGISTRY[mime]()            # look up class, instantiate

print(get_parser("application/pdf").parse("xx"))
# BaseParser()  # uncomment -> TypeError: can't instantiate abstract class
```

```python
# String Enum: typo-proof named states that are also strings
import enum

class JobStatus(str, enum.Enum):
    pending = "pending"
    completed = "completed"
    failed = "failed"

print(JobStatus.completed, "|", JobStatus.completed.value,
      "|", JobStatus.completed == "completed")
```

```python
# @dataclass: auto __init__/__repr__. Recursive via default_factory.
from dataclasses import dataclass, field

@dataclass
class HeadingNode:
    level: int
    title: str
    children: list = field(default_factory=list)   # safe mutable default

root = HeadingNode(1, "Doc", [HeadingNode(2, "Intro")])
print(root)
```

## 10. Control flow

```python
def heading_level(size, body, bold, text):
    if size > body + 1 and len(text) < 120:
        return 1
    elif bold and len(text) < 50:
        return 2
    elif text.isupper() and len(text) < 120:
        return 1
    return None

print(heading_level(16, 11, False, "BIG TITLE"))
```

```python
# guard clause + ternary
def title(headings):
    return headings[0] if headings else None

print(title(["A"]), title([]))
```

```python
# while loop with a manual counter
i, tokens, caps = 0, "a B C d".split(), []
while i < len(tokens):
    if tokens[i].isupper():
        caps.append(tokens[i])
    i += 1
print(caps)
```

```python
# try / except (bind the exception with `as`)
try:
    raise ValueError("not found")
except ValueError as exc:
    print("handled:", exc)
```

```python
# with: a custom context manager shows setup + guaranteed cleanup
class Resource:
    def __enter__(self):
        print("open"); return self
    def __exit__(self, *exc):
        print("close")

with Resource():
    print("using it")
```

```python
# match / case (Python 3.10+), mirrors your vector-store factory
def make_store(backend):
    match backend:
        case "pgvector": return "PgVectorStore"
        case "qdrant":   return "QdrantVectorStore"
        case other:      raise ValueError(f"unknown: {other}")

print(make_store("qdrant"))
```

## 16. Tie it together: a mini `extract_keywords`
A pure-stdlib version of the real `extract_keywords` (`search_service.py:43`).
It uses sets, slicing, string methods, comprehensions, and control flow together.

```python
STOP = frozenset({"what", "at", "has", "the", "a", "to", "of", "done", "projects"})

def extract_keywords(query):
    out = []
    for i, tok in enumerate(query.split()):
        w = tok.strip("?.!,;:\"'")          # clean punctuation
        if not w:
            continue
        # capitalised, not a stop word, not the first word -> likely an entity
        if w[0].isupper() and w.lower() not in STOP and i != 0:
            out.append(w)
    return out

print(extract_keywords("what projects has Amin done at Infosapex"))
print(extract_keywords("tell me about Python and FastAPI"))
```

## 19. Java vs Python: datatypes
Two ideas explain everything else:
1. **Java is statically typed** (types decided at *compile time*); **Python is dynamically typed** (the type lives on the **value**, decided at *runtime*).
2. **Java has primitives** (`int`, `double`, `char`, `boolean` are not objects); in **Python everything is an object** — even an integer.

The cells below demonstrate the consequences. (Java equivalents are in the comments.)

```python
# A) Dynamic typing: a variable is just a LABEL; it can re-point to any type.
x = 5
print(type(x).__name__, x)
x = "hello"          # Java: int x = 5; x = "hello";  -> COMPILE ERROR
print(type(x).__name__, x)
x = [1, 2, 3]
print(type(x).__name__, x)
```

```python
# B) Everything is an object — even an int has methods.
n = 5
print("bit_length:", n.bit_length())
print("to_bytes  :", (255).to_bytes(2, "big"))
print("types     :", type(5), type("a"), type(True))
# Java: int/double/char/boolean are primitives with NO methods (you box into Integer, etc.)
```

```python
# C) No overflow: Python int has unlimited (arbitrary) precision.
big = 2_000_000_000
print("big + big :", big + big)   # Java int would wrap to a NEGATIVE number
print("2 ** 100  :", 2 ** 100)    # huge, still fine
```

```python
# D) bool is a SUBCLASS of int in Python.
print("True == 1 :", True == 1)
print("True+True :", True + True)            # -> 2 (!)
print("isinstance:", isinstance(True, int))  # True
```

```python
# E) == compares VALUES; is compares IDENTITY (~ Java's ==). Use `is` only for None.
a = "py" + "thon"
b = "python"
print("a == b   :", a == b)   # True  -> value equality (use this)
print("a is b   :", a is b)   # identity — do NOT rely on it for values
val = None
print("val is None:", val is None)   # the correct way to test None
```

```python
# F) Duck typing: Python checks at RUNTIME whether the object can do the operation.
def greet(name):
    return name.upper()       # works for anything that has .upper()

print(greet("amin"))          # str -> fine
try:
    greet(42)                 # int has no .upper()
except AttributeError as e:
    print("runtime error:", e)
# Java rejects greet(42) at COMPILE time; Python only fails when the line runs.
```

```python
# G) A Python list holds ANY mix of types (Java List<T> forces one element type).
mixed = [1, "hi", 3.0, None, [2, 3]]
print(mixed)
print([type(v).__name__ for v in mixed])
```

**Summary**

| | Java | Python |
|---|---|---|
| Type is on the… | variable (declared) | value (runtime) |
| Basic values | primitives (`int`, `char`…) | objects (everything) |
| Wrong type caught | compile time | runtime |
| `int` size | fixed (overflows) | unlimited |
| `==` | references (objects) | values |

*Java — the variable has a fixed, compiler-checked type, and basic values are non-object primitives. Python — the value has the type, checked only at runtime, and everything is an object.*
